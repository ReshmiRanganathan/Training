```python
from delta.tables import DeltaTable
target_table = DeltaTable.forName(spark,"visits_target")
target_table.alias("target").merge(daily_visits_df.alias("source"),"target.visit_id = source.visit_id").whenMatchedUpdate(set={"patient_id":"source.patient_id","doctor_id":"source.doctor_id","visit_date":"source.visit_date","visit_status":"source.visit_status","tests_count":"source.tests_count"}).whenNotMatchedInsert(values={"visit_id":"source.visit_id","patient_id":"source.patient_id","doctor_id":"source.doctor_id","visit_date":"source.visit_date","visit_status":"source.visit_status","tests_count":"source.tests_count"}).execute()
```

MERGE is used for incremental loads because it can perform both update and insert operations in a single command. Existing records are updated and new records are inserted efficiently without creating duplicates. MERGE improves ETL performance and supports real-time data processing in Delta Lake.

Parquet is a storage format mainly used for efficient reading and writing of big data files. Delta Lake provides additional features on top of Parquet such as ACID transactions, schema enforcement, version history, time travel, update, delete, and merge operations.

```python
spark.sql("UPDATE patients_delta SET membership='Gold' WHERE patient_id=1002")
```

Update is easier in Delta Lake because Delta supports ACID transactions and maintains transaction logs. Raw Parquet files do not support direct update or delete operations efficiently. Delta Lake allows update, delete, merge, time travel, and rollback features.
