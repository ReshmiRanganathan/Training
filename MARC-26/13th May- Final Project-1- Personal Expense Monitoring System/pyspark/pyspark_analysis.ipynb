```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, col

# Create Spark session
spark = SparkSession.builder \
    .appName("Expense Monitoring System") \
    .getOrCreate()

# Load sample data
data = [
    (1, "Food", 250),
    (1, "Travel", 500),
    (2, "Shopping", 1200),
    (1, "Bills", 800),
    (2, "Food", 300)
]

columns = ["user_id", "category", "amount"]

df = spark.createDataFrame(data, columns)

print("Expense Data:")
df.show()

# Category-wise total
category_total = df.groupBy("category").agg(sum("amount").alias("total_expense"))

print("Category Wise Expense:")
category_total.show()

# High expense detection
high_expense = df.filter(col("amount") > 1000)

print("High Expense Transactions:")
high_expense.show()

spark.stop()
```
