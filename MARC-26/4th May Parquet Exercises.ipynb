# Parquet Data Processing using PySpark

This notebook demonstrates reading, writing, partitioning, and converting Parquet data to Delta Lake using Azure Databricks in hybrid mode.


```python
data = [
(101,"Arjun Reddy","Hyderabad","Cardiology",5000),
(102,"Sneha Kapoor","Delhi","Orthopedics",3000),
(103,"Rahul Sharma","Mumbai","Dermatology",1500),
(104,"Priya Nair","Bangalore","Cardiology",5000),
(105,"Vikram Singh","Chennai","Neurology",7000)
]

columns = ["visit_id","patient_name","city","department","consultation_fee"]

df = spark.createDataFrame(data, columns)
display(df)
```

```python
df.write.mode("overwrite").parquet("/tmp/patient_parquet")
```

```python
parquet_df = spark.read.parquet("/tmp/patient_parquet")
display(parquet_df)
```

```python
parquet_df.printSchema()
```

```python
spark.read.parquet("/tmp/patient_parquet").select("patient_name","city").show()
```

```python
spark.read.parquet("/tmp/patient_parquet").filter("consultation_fee > 3000").show()
```

```python
df.write.mode("overwrite").partitionBy("city").parquet("/tmp/patient_parquet_partitioned")
```

```python
spark.read.parquet("/tmp/patient_parquet_partitioned").show()
```

```python
spark.read.parquet("/tmp/patient_parquet_partitioned").filter("city = 'Hyderabad'").show()
```

### Partition Pruning

Only the required partition (Hyderabad) is read, improving performance.


```python
new_data = [
(106,"Ananya Das","Kolkata","Orthopedics",3000)
]
new_df = spark.createDataFrame(new_data, columns)
new_df.write.mode("append").parquet("/tmp/patient_parquet")
```

```python
df.write.mode("overwrite").parquet("/tmp/patient_parquet")
```

```python
%sql
SELECT * FROM parquet.`/tmp/patient_parquet`;
```

```python
%sql
CONVERT TO DELTA parquet.`/tmp/patient_parquet`;
```

```python
%sql
SELECT * FROM delta.`/tmp/patient_parquet`;
```

```python
%sql
UPDATE delta.`/tmp/patient_parquet` SET consultation_fee = 6000 WHERE visit_id = 101;
```

```python
%sql
SELECT * FROM delta.`/tmp/patient_parquet`;
```

Parquet is immutable and does not support UPDATE operations.
After converting to Delta, UPDATE works because Delta supports ACID transactions.

Due to environment constraints, Parquet data is queried directly using SQL instead of creating a table.
