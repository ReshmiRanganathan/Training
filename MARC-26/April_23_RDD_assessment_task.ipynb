```python
from pyspark.sql import SparkSession
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("RDD_Task") \
    .getOrCreate()
sc = spark.sparkContext
print("Spark Ready")
```

```python
logs = [
"Rahul login",
"Sneha login",
"Arjun login",
"Rahul purchase",
"Priya login",
"Sneha logout",
"Rahul login",
"Karan login",
"Arjun purchase",
"Sneha login",
"Rahul logout",
"Amit login",
"Priya purchase",
"Karan purchase",
"Sneha login",
"Rahul purchase",
"Meera login",
"Arjun login",
"Sneha purchase",
"Rahul login"
]
logs_rdd = sc.parallelize(logs)
print(logs_rdd.collect())
```

```python
sales = [
("Laptop",1,75000),
("Mouse",3,500),
("Keyboard",2,1500),
("Laptop",1,75000),
("Monitor",1,12000),
("Mouse",2,500),
("Keyboard",1,1500),
("Laptop",1,75000),
("Monitor",2,12000),
("Mouse",3,500),
("Laptop",1,75000),
("Keyboard",2,1500),
("Monitor",1,12000),
("Mouse",1,500),
("Laptop",1,75000)
]
sales_rdd = sc.parallelize(sales)
print(sales_rdd.collect())
```

```python
employees = [
("Rahul","IT",70000),
("Sneha","HR",60000),
("Arjun","IT",75000),
("Priya","Finance",80000),
("Karan","IT",50000),
("Amit","HR",58000),
("Meera","Finance",82000),
("Ravi","IT",72000),
("Neha","HR",61000),
("Vikram","Finance",90000)
]
emp_rdd = sc.parallelize(employees)
print(emp_rdd.collect())
```

```python
print("1.", logs_rdd.count())
print("2.", logs_rdd.take(5))
print("3.", logs_rdd.map(lambda x:x.split()[0]).collect())
print("4.", logs_rdd.map(lambda x:x.split()[1]).distinct().collect())
print("5.", logs_rdd.map(lambda x:x.split()[0]).distinct().collect())
print("6.", logs_rdd.filter(lambda x:x.split()[1]=="login").count())
print("7.", logs_rdd.filter(lambda x:x.split()[1]=="purchase").collect())
```

```python
sentences = sc.parallelize([
"big data spark",
"spark hadoop hive",
"python spark hadoop",
"spark streaming kafka"
])
print("8.", sentences.flatMap(lambda x:x.split()).collect())
print("9.", sentences.flatMap(lambda x:x.split()).distinct().collect())
print("10.", sentences.flatMap(lambda x:x.split()).map(lambda x:(x,1)).reduceByKey(lambda a,b:a+b).collect())
```

```python
print("11.", sales_rdd.map(lambda x:(x[0],x[1]*x[2])).collect())
print("12.", sales_rdd.map(lambda x:x[1]*x[2]).reduce(lambda a,b:a+b))
print("13.", sales_rdd.map(lambda x:x[0]).distinct().collect())
print("14.", sales_rdd.filter(lambda x:x[1]>1).collect())
print("15.", sales_rdd.map(lambda x:(x[0],x[1])).reduceByKey(lambda a,b:a+b).collect())
print("16.", sales_rdd.map(lambda x:(x[0],x[1]*x[2])).reduceByKey(lambda a,b:a+b).collect())
print("17.", sales_rdd.map(lambda x:(x[0],x[1]*x[2])).reduceByKey(lambda a,b:a+b).sortBy(lambda x:x[1],False).take(1))
```

```python
print("18.", emp_rdd.map(lambda x:x[0]).collect())
print("19.", emp_rdd.filter(lambda x:x[2]>70000).collect())
print("20.", emp_rdd.map(lambda x:x[1]).distinct().collect())
print("21.", emp_rdd.map(lambda x:(x[1],x[2])).reduceByKey(lambda a,b:a+b).collect())
print("22.", emp_rdd.map(lambda x:(x[1],(x[2],1))).reduceByKey(lambda a,b:(a[0]+b[0],a[1]+b[1])).mapValues(lambda x:x[0]/x[1]).collect())
print("23.", emp_rdd.sortBy(lambda x:x[2],False).take(1))
```

```python
print("24.", emp_rdd.map(lambda x:(x[1],1)).reduceByKey(lambda a,b:a+b).collect())
print("25.", emp_rdd.sortBy(lambda x:x[2],False).collect())
print("26.", emp_rdd.sortBy(lambda x:x[2],False).take(3))
print("27.", emp_rdd.map(lambda x:(x[1],x[0])).collect())
print("28.", emp_rdd.map(lambda x:(x[1],x[0])).groupByKey().mapValues(list).collect())
```

```python
print("29.", emp_rdd.map(lambda x:(x[1],x[2])).reduceByKey(lambda a,b:a+b).sortBy(lambda x:x[1],False).take(1))
print("30.", emp_rdd.map(lambda x:(x[1],(1,x[2]))).reduceByKey(lambda a,b:(a[0]+b[0],a[1]+b[1])).collect())
```

```python
revenue = sales_rdd.map(lambda x:(x[0],x[1]*x[2])).reduceByKey(lambda a,b:a+b)
print("Revenue Per Product")
print(revenue.sortBy(lambda x:x[1],False).collect())
print("Top Selling Product")
print(sales_rdd.map(lambda x:(x[0],x[1])).reduceByKey(lambda a,b:a+b).sortBy(lambda x:x[1],False).take(1))
print("Highest Revenue Product")
print(revenue.sortBy(lambda x:x[1],False).take(1))
```
