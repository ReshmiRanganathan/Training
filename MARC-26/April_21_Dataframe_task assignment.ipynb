```python
print("Part-18")
print("Ex:53")
from pyspark.sql.types import StringType
def category(x):
    if x > 50000:
        return "Premium"
    elif x >= 20000:
        return "Medium"
    else:
        return "Small"
cat_udf = udf(category, StringType())
sales.withColumn("sale_category", cat_udf(col("amount"))).show()
```

```python
print("part-20")
print("Ex:56")
import ipywidgets as widgets
dropdown = widgets.Dropdown(
    options=["IT", "HR", "Finance"],
    description="Dept:"
)
display(dropdown)
```

```python
print("Ex:57")
slider = widgets.IntSlider(
    min=40000,
    max=90000,
    step=1000,
    description="Salary:"
)
display(slider)
```

```python
print("Ex:58")
dept = dropdown.value
sal = slider.value
employees.filter(
    (col("department") == dept) &
    (col("salary") >= sal)
).show()
```

```python
print("Mini-Project")
from pyspark.sql.functions import sum, avg, col
top_employee = employees.orderBy(col("salary").desc()).first()
average_salary = employees.select(avg("salary")).first()[0]
dept_count = employees.groupBy("department").count().collect()
total_sales = sales.select(sum("amount")).first()[0]
best_sales_employee = sales.groupBy("emp_id") \
    .sum("amount") \
    .orderBy(col("sum(amount)").desc()) \
    .first()
with open("final_report.txt", "w") as file:
    file.write("FINAL REPORT\n")
    file.write("----------------------\n")
    file.write(f"Top Employee : {top_employee['name']} - {top_employee['salary']}\n")
    file.write(f"Average Salary : {average_salary}\n")
    file.write("\nDepartment Employee Count:\n")
    for row in dept_count:
        file.write(f"{row['department']} : {row['count']}\n")
    file.write(f"\nTotal Sales : {total_sales}\n")
    file.write(f"Best Sales Employee ID : {best_sales_employee['emp_id']}\n")
    file.write(f"Sales Amount : {best_sales_employee['sum(amount)']}\n")
print("final_report.txt created successfully")
from google.colab import files
files.download("final_report.txt")
```
