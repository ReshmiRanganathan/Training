```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum, col

# Create Spark session
spark = SparkSession.builder \
    .appName("Smart Home Energy Tracker") \
    .getOrCreate()

# Sample sensor data
data = [
    (1, "AC", 5.5),
    (2, "Refrigerator", 2.0),
    (3, "TV", 1.2),
    (1, "AC", 6.1),
    (2, "Refrigerator", 2.3),
    (3, "TV", 1.5)
]

columns = ["device_id", "device_name", "energy_kwh"]

df = spark.createDataFrame(data, columns)

print("Sensor Data:")
df.show()

# Total energy usage by device
device_summary = df.groupBy("device_name").agg(
    sum("energy_kwh").alias("total_energy")
)

print("Top Energy Consuming Devices:")
device_summary.show()

# High energy usage detection
high_usage = df.filter(col("energy_kwh") > 5)

print("High Usage Devices:")
high_usage.show()

spark.stop()
```
