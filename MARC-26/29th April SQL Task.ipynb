```python
%sql
CREATE OR REPLACE TABLE employee_salary_analysis AS SELECT emp_name, city, department, salary, experience, CASE WHEN salary>=95000 THEN 'High' WHEN salary>=80000 THEN 'Medium' ELSE 'Low' END AS salary_category, AVG(salary) OVER(PARTITION BY department) AS department_avg_salary FROM employees;
```
