```python
from pyspark.sql.functions import col
sales = sales.withColumn("revenue", col("quantity") * col("price"))
```

```python
final_df = emp.join(sales, on="emp_id", how="inner")
print("1. Joined Employees and Sales Data")
final_df.show(truncate=False)
```

```python
employee_revenue = final_df.groupBy("name").sum("revenue").withColumnRenamed("sum(revenue)", "Revenue")
print("2. Total Revenue Generated by Each Employee")
employee_revenue.show(truncate=False)
```

```python
print("3. Top 5 Employees by Sales")
top5 = employee_revenue.orderBy(col("Revenue").desc())
top5.show(5, truncate=False)
```

```python
print("4. Department Generating Highest Revenue")
dept_revenue = final_df.groupBy("department").sum("revenue").withColumnRenamed("sum(revenue)", "Revenue").orderBy(col("Revenue").desc())
dept_revenue.show(1, truncate=False)
```

```python
employee_revenue.coalesce(1).write.mode("overwrite").option("header", True).csv("final_sales_report.csv")
print("5. final_sales_report.csv saved successfully")
```
