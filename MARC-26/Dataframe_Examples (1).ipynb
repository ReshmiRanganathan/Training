```python
from pyspark.sql import SparkSession
spark=SparkSession.builder.appName("Dataframe Examples").getOrCreate()
print("Spark Session Created")
```

```python
data=[(1,"Rahul","IT",70000,"2024-01-10"),(2,"Sneha","HR",60000,"2024-02-15"),(3,"Arjun","IT",75000,"2024-03-01"),(4,"Priya","Finance",80000,"2024-01-25"),(5,"Karan",None,50000,"2024-02-20")]
columns=["emp_id","name","department","salary","joining_date"]
df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.show()
df.printSchema()
df.columns
df.count()
```

```python
df.show(3)
```

```python
df_renamed=df.withColumnRenamed("salary","emp_salary")
df_renamed.show()
```

```python
df.filter(df.department=="IT").show()
```

```python
df.filter((df.salary>60000)&(df.department=="IT")).show()
```

```python
df.drop("joining_date").show()
```

```python
from pyspark.sql.functions import sum,avg,max,min
df.select(sum("salary").alias("total_salary"),avg("salary").alias("avg_salary"),max("salary").alias("max_salary"),min("salary").alias("min_salary")).show()
```

```python
dept_data = [
    ("IT", "Bangalore"),
    ("HR", "Mumbai"),
    ("Finance", "Delhi")
]

dept_columns = ["department", "location"]

dept_df = spark.createDataFrame(dept_data, dept_columns)
dept_df.show()
```

```python
df.join(dept_df,on="department",how="inner").show()
```

```python
df.join(dept_df,on="department",how="left").show()
```

```python
df.join(dept_df,on="department",how="right").show()
```

```python
df.withColumn("bonus",df.salary*0.10).show()
```

```python
from pyspark.sql.functions import col
df.filter(col("department").isNull()).show()
```

```python
df.na.fill({"department":"unknown"}).show()                    #na refers to null, here in place of null we give unknown
```

```python
df.na.drop().show()      #dropping the null record
```

```python
from pyspark.sql.functions import upper,lower,length
df.select("name",upper("name").alias("upper_name"),lower("name").alias("lower_name"),length("name").alias("length_name")).show()
```

```python
df=spark.read.csv("employees.csv",header=True,inferSchema=True)
df.show()
```

```python
from pyspark.sql.functions import to_date, year, month, dayofmonth
df_dates=df.withColumn("joining_date",to_date("joining_date"))
df_dates.show()
```

```python
data = [
    (1, "Rahul", "IT", 70000, "2024-01-10"),
    (2, "Sneha", "HR", 60000, "2024-02-15"),
    (3, "Arjun", "IT", 75000, "2024-03-01"),
    (4, "Priya", "Finance", 80000, "2024-01-25"),
    (5, "Karan", None, 50000, "2024-02-20")
]

columns = ["emp_id", "name", "department", "salary", "joining_date"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
from pyspark.sql.functions import to_date,year,month,dayofmonth
df.dates=df.withColumn("joining_date",to_date("joining_date"))
df_dates.show()
```

```python
df_dates.select("name","joining_date",year("joining_date").alias("year"),month("joining_date").alias("month"),dayofmonth("joining_date").alias("day")).show()
```
